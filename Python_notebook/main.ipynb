```python
import yfinance as yf
import pandas as pd
import numpy as np

stocks = {
    'TCS': ('TCS.NS', 'IT'),
    'Infosys': ('INFY.NS', 'IT'),
    'HDFC Bank': ('HDFCBANK.NS', 'Banking'),
    'ICICI Bank': ('ICICIBANK.NS', 'Banking'),
    'Reliance': ('RELIANCE.NS', 'Energy'),
    'ONGC': ('ONGC.NS', 'Energy'),
    'HUL': ('HINDUNILVR.NS', 'FMCG'),
    'ITC': ('ITC.NS', 'FMCG'),
    'Sun Pharma': ('SUNPHARMA.NS', 'Pharma'),
    'Dr Reddys': ('DRREDDY.NS', 'Pharma')
}

all_data = []

for name, (ticker, sector) in stocks.items():
    print(f"Downloading {name}...")
    df = yf.download(ticker, start="2020-01-01", end="2024-12-31", auto_adjust=True)
    
    # Flatten any MultiIndex columns
    df.columns = df.columns.get_level_values(0)
    
    # Keep only OHLCV columns
    df = df[['Open', 'High', 'Low', 'Close', 'Volume']].copy()
    
    # Add metadata
    df['Stock'] = name
    df['Ticker'] = ticker
    df['Sector'] = sector
    
    all_data.append(df)

# Stack vertically
df_raw = pd.concat(all_data, axis=0)
df_raw.reset_index(inplace=True)
df_raw.rename(columns={'index': 'Date'}, inplace=True)

print(f"\n✅ Total records: {len(df_raw)}")
print(f"✅ Columns: {df_raw.columns.tolist()}")
print(df_raw.head(10))




```

```python
# we can clarliy see 6    ONGC    2.189953 hihly risky beacuise the std is high 

# now we will see the growth of stocks

Stock_Growth = df_raw.groupby('Stock')['Close'].agg(
    First_Price = 'first',
    last_Price = 'last'
).reset_index()

Stock_Growth['growth%'] = (Stock_Growth['last_Price'] - Stock_Growth['First_Price'])/Stock_Growth['First_Price']*100
Stock_Growth.sort_values(by='growth%', ascending=False)
```

```python
# IT vs Banking vs Energy vs FMCG vs Pharma — kaun sa sector best performer?
best_performer = df_raw.groupby('Sector')['Close'].agg(
    first_price = 'first',
    last_price = 'last'
).reset_index()

best_performer['growth%'] = (best_performer['last_price'] - best_performer['first_price'])/best_performer['first_price']*100
best_performer.sort_values(by='growth%',ascending=False)
```
